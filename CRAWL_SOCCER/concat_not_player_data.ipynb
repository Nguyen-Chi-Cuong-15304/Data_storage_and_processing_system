```python
import pandas as pd
import numpy as np

Not_GK_Players_Data = 'Not_GK_Players_Data.csv'
Not_GK_Players_Stat = 'Not_GK_Players_Stat.csv'

soccer = pd.read_csv(Not_GK_Players_Data)
stat = pd.read_csv(Not_GK_Players_Stat)

soccer_data = pd.concat([soccer, stat], axis=1)

Soccer_Data_Finals = 'Soccer_Data_Finals.csv'

# soccer_data.to_csv(Soccer_Data_Finals, index=False)

# print(soccer_data.head())
```

```python
# đếm số dòng bị có cột bị null
print(soccer_data.isnull().sum())

# chuyển các dòng có cột player_agent bị null thành None
soccer_data['player_agent'] = soccer_data['player_agent'].astype(str).str.strip() # Chuyển sang string và loại bỏ khoảng trắng
soccer_data['player_agent'] = soccer_data['player_agent'].replace('', 'NoInfo') # Thay thế chuỗi rỗng bằng None
soccer_data['player_agent'] = soccer_data['player_agent'].replace(np.nan, 'NoInfo')
print('--'*50)

print (f'Có {soccer_data["player_agent"].isnull().sum()} dòng có cột player_agent bị null')
#In ra tổng số dòng có cột bất kỳ bị null
print (f'Tổng số dòng có cột bất kỳ bị null là {soccer_data.isnull().sum()}')




print('--'*50)
print("\nSố lượng null trên mỗi cột:")
print(soccer_data.isnull().sum())

# --- Cách đếm tổng số dòng có ít nhất một cột bị null ---

# Bước 1: Sử dụng isnull() để có DataFrame True/False về null
# True ở những vị trí có null, False ở những vị trí không null
isnull_df = soccer_data.isnull()
# print("\nDataFrame isnull():")
# print(isnull_df)

# Bước 2: Sử dụng .any(axis=1) để kiểm tra xem có ít nhất một True trên MỖI DÒNG hay không
# axis=1 chỉ định kiểm tra theo chiều ngang (qua các cột)
# Kết quả là một Series boolean, True nếu dòng đó có ít nhất một null
rows_with_any_null = isnull_df.any(axis=1)
# print("\nSeries boolean (rows with any null):")
# print(rows_with_any_null)

# Bước 3: Sử dụng .sum() trên Series boolean để đếm số lượng True
# Tổng số True là tổng số dòng có ít nhất một null
total_rows_with_any_null = rows_with_any_null.sum()

print(f"\nTổng số dòng có ít nhất một cột bị null: {total_rows_with_any_null}")

# --- So sánh với tổng số dòng ---
total_rows = len(soccer_data) # Hoặc soccer_data.shape[0]
print(f"Tổng số dòng trong DataFrame: {total_rows}")

# Số dòng không có null
rows_with_no_null = total_rows - total_rows_with_any_null
print(f"Tổng số dòng KHÔNG có cột nào bị null: {rows_with_no_null}")
```

```python
# data_to_clean = pd.read_csv(Soccer_Data_Finals)
data_to_clean = soccer_data.copy()
cleaned_data = data_to_clean.dropna(how='any', axis=0)
cleaned_data = cleaned_data.drop_duplicates()

valid_values = ['left', 'right', 'both']
# Loại bỏ các hàng có giá trị không hợp lệ trong cột 'strong_foot'
cleaned_data = cleaned_data[cleaned_data['strong_foot'].isin(valid_values)]


Cleaned_Data = 'Cleaned_Data.csv'
cleaned_data.to_csv(Cleaned_Data, index=False)
#In ra header của cleaned_data và kiểu dữ liệu của từng cột

print("Header of cleaned data:")
print(cleaned_data.columns)
print("\nData types of each column:")
print(cleaned_data.dtypes)

#Đếm số cột có player_agent là 'NoInfo'
no_info_count = cleaned_data[cleaned_data['player_agent'] == 'NoInfo'].shape[0]
print(f"\nSố lượng cột có player_agent là 'NoInfo': {no_info_count}")
```
